```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_data(filename):
    data = np.loadtxt(filename)
    return data[:,0], data[:,1]
```

```python
def exact(x, t):
    return np.sin(2 * np.pi * (x - t))
```

```python
v = 0.70
times = [0, 0.25, 0.5, 0.75, 1.0]
```

```python
plt.figure(figsize=(12,6))

for t in times:
    x, u = load_data(f"lf_{t}_v_{v:.2f}.dat")
    plt.plot(x, u, label=f"LF t={t}")
    plt.plot(x, exact(x, t), '--', label=f"Exact t={t}")

plt.title("Lax–Friedrich Scheme")
plt.grid()
plt.legend()
plt.show()
```

```python
data = np.loadtxt("lf_dx_vs_error.dat")

dx = data[:,0]
error = data[:,1]

plt.figure(figsize=(8,6))

plt.loglog(dx, error, 'o-', label="L2 Error")

plt.xlabel("Δx")
plt.ylabel("L2 Error")
plt.title("Convergence of Lax–Friedrichs Scheme")

plt.show()

log_dx = np.log(dx)
log_err = np.log(error)

coeffs = np.polyfit(log_dx, log_err, 1)
rate = coeffs[0]

print(f"Observed convergence rate = {rate:.4f}")
```
