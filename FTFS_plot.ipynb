```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_data(filename):
    data = np.loadtxt(filename)
    return data[:,0], data[:,1]
```

```python
def exact(x, t):
    return np.sin(2 * np.pi * (x - t))
```

```python
times = [0, 0.25, 0.5, 0.75, 1.0]
```

```python
plt.figure(figsize=(14,8))

plt.subplot(2,1,1)
for t in times:
    x, u = load_data(f"ftbs_{t}.dat")
    plt.plot(x, u, label=f"FTBS t={t}")
    plt.plot(x, exact(x, t), '--', label=f"Exact t={t}")

plt.title("FTBS Normal")
plt.grid()
plt.legend()

plt.subplot(2,1,2)
for t in times:
    x, u = load_data(f"ftbs_{t}.dat")
    plt.plot(x, u, label=f"FTBS t={t}")
    plt.plot(x, exact(x, t), '--', label=f"Exact t={t}")

plt.ylim(-1.5,1.5)
plt.title("FTFBS Zoomed")
plt.grid()
plt.legend()

plt.show()
```
